и результатами их команд в сезоне 2025/26.

## Задачи первичного анализа

1. Изучить структуру датасета.
2. Проверить типы данных.
3. Найти пропущенные значения и дубликаты.
4. Выделить матчи регулярного сезона.
5. Определить дальнейшие аналитические метрики.

## Первые наблюдения

Заполним после выполнения проверок.

```python
from pathlib import Path
import pandas as pd

# Путь к папке проекта
PROJECT_ROOT = Path.cwd().resolve()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "nba_box_scores_2025_26.csv"
)

df = pd.read_csv(DATA_PATH)

print("Размер датасета:", df.shape)
display(df.head())
```

```python
print("Названия столбцов:")
print(df.columns.tolist())

print("\nИнформация о датасете:")
df.info()

print("\nЧисловая статистика:")
display(df.describe())
```

```python
# Количество пропусков в каждом столбце
missing_values = (
    df.isna()
    .sum()
    .sort_values(ascending=False)
)

print("Пропущенные значения:")
display(
    missing_values[missing_values > 0].to_frame("missing_count")
)

# Полностью повторяющиеся строки
print("Полных дубликатов:", df.duplicated().sum())

# Повторение ключа игрок + матч
duplicate_keys = df.duplicated(
    subset=["game_id", "player_id"]
)

print(
    "Повторов по game_id + player_id:",
    duplicate_keys.sum()
)

if duplicate_keys.any():
    display(
        df.loc[
            duplicate_keys,
            ["game_id", "player_id", "player"]
        ].head(10)
    )
```

```python
print("Типы матчей:")
display(df["season_type"].value_counts(dropna=False))

print("\nСтатусы участия:")
display(df["did_not_play"].value_counts(dropna=False))

print("\nКоличество уникальных объектов:")
print("Игроков:", df["player_id"].nunique())
print("Команд:", df["team"].nunique())
print("Матчей:", df["game_id"].nunique())
```

```python
regular_season = df.loc[
    df["season_type"] == "Regular Season"
].copy()

print("Все строки:", len(df))
print("Регулярный сезон:", len(regular_season))

display(
    regular_season[
        [
            "game_date",
            "player",
            "team",
            "opponent",
            "pts",
            "ast",
            "trb",
        ]
    ].head(10)
)
```

```python
# 1. Строки без имени или идентификатора игрока
missing_players = df[
    df["player"].isna() | df["player_id"].isna()
]

print("Строк без имени или ID:", len(missing_players))

display(
    missing_players[
        [
            "game_id",
            "game_date",
            "player",
            "player_id",
            "team",
            "did_not_play",
            "minutes",
            "pts",
        ]
    ].head(40)
)

# 2. Строки без минут, но со статистикой
missing_minutes = df[
    df["minutes"].isna() & df["pts"].notna()
]

print("\nСтрок без minutes, но с pts:", len(missing_minutes))

display(
    missing_minutes[
        [
            "game_id",
            "game_date",
            "player",
            "team",
            "did_not_play",
            "minutes",
            "pts",
            "ast",
        ]
    ].head(20)
)

# 3. Проверка связи пропусков с неучастием
print("\nСтатусы did_not_play при отсутствии pts:")

display(
    df.loc[
        df["pts"].isna(),
        "did_not_play"
    ].value_counts(dropna=False)
)
```

При первичной проверке данных NBA за сезон 2025/26 были обнаружены пропуски в статистике игроков. Большинство из них предположительно связано с отсутствием игроков на площадке. Дополнительно выявлены 33 записи без имени и идентификатора баскетболиста, а также случаи, когда игровое время отсутствует при нулевых статистических показателях и флаге did_not_play = False. Эти особенности требуют отдельной обработки перед расчётом аналитических метрик.

```python
# Проверка проблемных записей

# 1. К каким командам относятся строки без ID?
print("Команды с отсутствующими player_id:")
display(
    df.loc[
        df["player_id"].isna(),
        "team"
    ].value_counts(dropna=False)
)

# 2. Проверяем, есть ли статистика у игроков без минут
stats_columns = [
    "pts", "ast", "trb", "stl", "blk", "tov"
]

missing_minutes = df[
    df["minutes"].isna()
    & df["pts"].notna()
].copy()

print("\nКоличество записей без минут, но с очками:")
print(len(missing_minutes))

print("\nМаксимальные значения показателей:")
display(missing_minutes[stats_columns].max())

# 3. Проверяем статусы участия при отсутствии минут
print("\nСтатусы did_not_play для таких записей:")
display(
    missing_minutes["did_not_play"].value_counts(
        dropna=False
    )
)

# 4. Проверяем значения minutes у остальных игроков
print("\nПримеры значений игрового времени:")
display(df["minutes"].dropna().head(15))
```
